# Plotting a synthetic NEXAFS-like spectrum

This is an **authoring/rendering demonstration**, not a NEXAFSFit fit. It uses an illustrative height-based Gaussian function rather than the engine's integrated-area API. For a tested native fit using the supplied numerical package, follow [A native NEXAFSFit single-spectrum fit](single-spectrum-fit.ipynb).

## Create a small, deterministic example

We construct two overlapping Gaussian components and a sloping background. The numerical parameters here are arbitrary and are provided solely for documentation practice.

```python
import numpy as np
import matplotlib.pyplot as plt

energy = np.linspace(280, 300, 401)

def gaussian(x, centre, sigma, height):
    return height * np.exp(-0.5 * ((x - centre) / sigma) ** 2)

peak_a = gaussian(energy, 286.8, 0.35, 0.28)
peak_b = gaussian(energy, 288.0, 0.70, 0.17)
baseline = 0.03 + 0.002 * (energy - energy.min())
spectrum = baseline + peak_a + peak_b
```

```python
fig, ax = plt.subplots(figsize=(7, 3.6))
ax.plot(energy, spectrum, label="Synthetic spectrum", color="black", lw=1.8)
ax.plot(energy, baseline + peak_a, "--", label="Example component A")
ax.plot(energy, baseline + peak_b, "--", label="Example component B")
ax.set(xlabel="Photon energy (eV)", ylabel="Arbitrary intensity")
ax.legend(frameon=False)
fig.tight_layout()
plt.show()
```
